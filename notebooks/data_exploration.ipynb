```python
# Vidimo da su nedostajuce informacije u datasetovima popunjene sa '\N' pa ih odmah pri citanju menjamo sa oznakom za nedostajucu vrednost - NaN 

drivers = pd.read_csv("../data/raw/drivers.csv", na_values=["\\N"])
constructors = pd.read_csv("../data/raw/constructors.csv", na_values=["\\N"])
races = pd.read_csv("../data/raw/races.csv", na_values=["\\N"])
results = pd.read_csv("../data/raw/results.csv", na_values=["\\N"])
qualifying = pd.read_csv("../data/raw/qualifying.csv", na_values=["\\N"])
status = pd.read_csv("../data/raw/status.csv", na_values=["\\N"])
```

## Missing values

```python
for name, df in datasets.items():
    print("\nMissing values in " + name + " dataset, per column: ")
    print(df.isnull().sum())
```

```python
missing_values = {}

for name, df in datasets.items():
    missing_values[name] = {}

    for column in df.columns:
        count = df[column].isnull().sum()
        percentage = count / len(df) * 100

        if count > 0:
            missing_values[name][column] = (count, round(percentage, 2))
```

```python
for name, columns in missing_values.items():
    print(name + ":")

    if len(columns) == 0:
        print("  No missing values")
    else:
        for column, values in columns.items():
            print(" ", column, ":", values[0], "missing values,", values[1], "%")

    print()
```

```python
for name, df in datasets.items():
    count = df.duplicated().sum()
    print(name + ":", count, "duplicate rows")
```

```python
primary_keys = {
    "drivers": "driverId",
    "constructors": "constructorId",
    "races": "raceId",
    "results": "resultId",
    "quali": "qualifyId",
    "status": "statusId"
}

for name, df in datasets.items():
    key = primary_keys[name]
    count = df[key].duplicated().sum()
    print(name + ":", count, "duplicate IDs")
```
